## Our Geotechnical Problem

A borehole log contains **multiple soil layers**, each with its own properties:

```
BH-01 Borehole Log
──────────────────────────
Depth   N60   Soil           γ (kN/m³)
 1.5      5   Fill           16.0
 3.0     12   Silty Sand     17.5
 6.0     22   Sand           18.0
 9.0      8   Soft Clay      16.5
12.0     35   Dense Sand     19.0
15.0     55   Gravel         20.0
```

Without OOP, we'd track each property in **separate lists** — messy and error-prone.
With OOP, each layer is **one object** that carries all its data and behaviors together.

### The Blueprint Analogy

```
Class = Blueprint                Object = Actual thing
────────────────────         ────────────────────
SoilLayer class:                layer_1 = SoilLayer(...):
  - has a depth                   depth = 3.0
  - has an N60 value              N60 = 12
  - has a soil type               soil_type = "Silty Sand"
  - can classify itself           classify_consistency() → "Loose"
```

## Part 1: Class vs Object — The Simplest Example

A **class** is a template. An **object** (or instance) is one specific thing created from that template.

```python
class SoilLayer:        # ← The blueprint (class)
    pass

layer = SoilLayer()     # ← One actual layer (object)
```

You write the class **once**, then create as many objects as you need — one per soil layer in your borehole.

````python
# ─────────────────────────────────────────────────────
# SIMPLEST CLASS — Just to see the concept
# ─────────────────────────────────────────────────────

class SoilSample:
    """A minimal class — just to show the syntax."""
    pass

# Create two objects (instances) from the same class
sample_a = SoilSample()
sample_b = SoilSample()

# They are different objects
print(f"sample_a: {sample_a}")
print(f"sample_b: {sample_b}")
print(f"Same object? {sample_a is sample_b}")
print(f"Both are SoilSample? {isinstance(sample_a, SoilSample)}, {isinstance(sample_b, SoilSample)}")
print()
print("One class, two objects — like one Shelby tube design, many actual tubes.")
````

## Part 2: `__init__` — The Constructor

The `__init__` method runs automatically when you create an object. It sets up the object's data (called **attributes**).

```python
class SoilLayer:
    def __init__(self, soil_type, depth, N60, unit_weight):
        self.soil_type   = soil_type     # store on the object
        self.depth       = depth
        self.N60         = N60
        self.unit_weight = unit_weight
```

- `self` = "this specific object" (Python fills it in automatically)
- `self.depth = depth` means "save `depth` on THIS object so we can use it later"

````python
# ─────────────────────────────────────────────────────
# __init__ — Store soil properties on each object
# ─────────────────────────────────────────────────────

class SoilLayer:
    """Represents one soil layer in a borehole log."""
    
    def __init__(self, soil_type, depth, N60, unit_weight):
        self.soil_type   = soil_type      # e.g., "Sand", "Clay"
        self.depth       = depth          # meters below ground
        self.N60         = N60            # corrected SPT blow count
        self.unit_weight = unit_weight    # kN/m³

# Create objects — each one stores its own data
layer_1 = SoilLayer("Fill",       1.5,  5, 16.0)
layer_2 = SoilLayer("Silty Sand", 3.0, 12, 17.5)
layer_3 = SoilLayer("Sand",       6.0, 22, 18.0)

# Access attributes with dot notation
print(f"Layer 1: {layer_1.soil_type} at {layer_1.depth}m, N60={layer_1.N60}, \u03b3={layer_1.unit_weight} kN/m\u00b3")
print(f"Layer 2: {layer_2.soil_type} at {layer_2.depth}m, N60={layer_2.N60}, \u03b3={layer_2.unit_weight} kN/m\u00b3")
print(f"Layer 3: {layer_3.soil_type} at {layer_3.depth}m, N60={layer_3.N60}, \u03b3={layer_3.unit_weight} kN/m\u00b3")
print()
print("Each object carries its own data — no parallel lists needed!")
````

## Part 3: Instance Methods — Objects That DO Things

Methods are functions that belong to a class. They can access the object's data via `self`.

```python
class SoilLayer:
    def classify_consistency(self):
        if self.N60 < 4:
            return "Very Loose"
        ...
```

Now `layer.classify_consistency()` uses THAT layer's N60 value automatically.

````python
# ─────────────────────────────────────────────────────
# INSTANCE METHODS — Behaviors that use the object's data
# ─────────────────────────────────────────────────────

class SoilLayer:
    """Represents one soil layer in a borehole log."""
    
    def __init__(self, soil_type, depth, N60, unit_weight):
        self.soil_type   = soil_type
        self.depth       = depth          # m
        self.N60         = N60            # corrected SPT blow count
        self.unit_weight = unit_weight    # kN/m³
    
    def classify_consistency(self):
        """Classify based on SPT N60 (granular soils)."""
        if self.N60 < 4:   return "Very Loose"
        if self.N60 < 10:  return "Loose"
        if self.N60 < 30:  return "Medium Dense"
        if self.N60 < 50:  return "Dense"
        return "Very Dense"
    
    def estimate_vs(self):
        """Estimate shear wave velocity (Imai & Tonouchi, 1982): Vs = 97 * N^0.314."""
        if self.N60 <= 0:
            return 0.0
        return round(97 * (self.N60 ** 0.314), 1)
    
    def overburden_stress(self):
        """Total vertical stress at layer depth: sigma_v = gamma * depth."""
        return round(self.unit_weight * self.depth, 2)

# Create layers and call their methods
layers = [
    SoilLayer("Fill",       1.5,  5, 16.0),
    SoilLayer("Silty Sand", 3.0, 12, 17.5),
    SoilLayer("Sand",       6.0, 22, 18.0),
    SoilLayer("Soft Clay",  9.0,  8, 16.5),
    SoilLayer("Dense Sand",12.0, 35, 19.0),
]

print(f"{'Soil':14s} {'Depth':>5} {'N60':>4} {'Class':14s} {'Vs':>6} {'\u03c3v':>8}")
print("\u2500" * 56)
for layer in layers:
    print(f"{layer.soil_type:14s} {layer.depth:5.1f} {layer.N60:4d} "
          f"{layer.classify_consistency():14s} {layer.estimate_vs():6.1f} "
          f"{layer.overburden_stress():7.1f} kPa")
````

## Part 4: `__str__` and `__repr__` — Readable Output

By default, `print(layer)` shows something like `<SoilLayer object at 0x...>` — useless.

We fix this with two special methods:

| Method | Purpose | When It's Used |
|--------|---------|----------------|
| `__str__` | Human-readable string | `print(layer)`, `str(layer)` |
| `__repr__` | Developer/debug string | In the REPL, `repr(layer)`, inside lists |

````python
# ─────────────────────────────────────────────────────
# __str__ and __repr__ — Making objects print nicely
# ─────────────────────────────────────────────────────

class SoilLayer:
    """Represents one soil layer in a borehole log."""
    
    def __init__(self, soil_type, depth, N60, unit_weight):
        self.soil_type   = soil_type
        self.depth       = depth
        self.N60         = N60
        self.unit_weight = unit_weight
    
    def classify_consistency(self):
        if self.N60 < 4:   return "Very Loose"
        if self.N60 < 10:  return "Loose"
        if self.N60 < 30:  return "Medium Dense"
        if self.N60 < 50:  return "Dense"
        return "Very Dense"
    
    def estimate_vs(self):
        if self.N60 <= 0:
            return 0.0
        return round(97 * (self.N60 ** 0.314), 1)
    
    def overburden_stress(self):
        return round(self.unit_weight * self.depth, 2)
    
    def __str__(self):
        """Human-readable: used by print()."""
        return (f"{self.soil_type} at {self.depth}m | "
                f"N60={self.N60}, \u03b3={self.unit_weight} kN/m\u00b3 | "
                f"{self.classify_consistency()}")
    
    def __repr__(self):
        """Developer-readable: shows how to recreate the object."""
        return (f"SoilLayer('{self.soil_type}', {self.depth}, "
                f"{self.N60}, {self.unit_weight})")

# Now printing is clean
layer = SoilLayer("Sand", 6.0, 22, 18.0)

print("print(layer) uses __str__:")
print(f"  {layer}")
print()
print("repr(layer) uses __repr__:")
print(f"  {repr(layer)}")
print()

# __repr__ also shows inside lists
some_layers = [
    SoilLayer("Fill", 1.5, 5, 16.0),
    SoilLayer("Sand", 6.0, 22, 18.0),
]
print("Inside a list (uses __repr__):")
print(f"  {some_layers}")
````

## Part 5: Class Attributes vs Instance Attributes

- **Instance attributes** (`self.depth`) — unique to each object
- **Class attributes** (defined outside `__init__`) — shared by ALL objects of that class

```python
class SoilLayer:
    gravity = 9.81          # ← Class attribute: same for ALL layers
    
    def __init__(self, ...):
        self.depth = depth  # ← Instance attribute: different per layer
```

Perfect for physical constants like gravity, water density, or atmospheric pressure.

````python
# ─────────────────────────────────────────────────────
# CLASS ATTRIBUTES — Shared constants across all instances
# ─────────────────────────────────────────────────────

class SoilLayer:
    """Soil layer with shared physical constants."""
    
    # ---- Class attributes (shared by ALL instances) ----
    gravity        = 9.81    # m/s²
    water_density  = 9.81    # kN/m³
    atm_pressure   = 101.3   # kPa
    layer_count    = 0       # track how many layers we create
    
    def __init__(self, soil_type, depth, N60, unit_weight):
        # ---- Instance attributes (unique to each object) ----
        self.soil_type   = soil_type
        self.depth       = depth
        self.N60         = N60
        self.unit_weight = unit_weight
        
        # Increment the shared counter
        SoilLayer.layer_count += 1
    
    def effective_unit_weight(self, water_table):
        """Submerged unit weight below water table."""
        if self.depth > water_table:
            return round(self.unit_weight - self.water_density, 2)
        return self.unit_weight
    
    def __str__(self):
        return f"{self.soil_type} at {self.depth}m | N60={self.N60}"

# Create layers
l1 = SoilLayer("Fill",       1.5,  5, 16.0)
l2 = SoilLayer("Silty Sand", 3.0, 12, 17.5)
l3 = SoilLayer("Sand",       6.0, 22, 18.0)

# Class attribute is shared
print("Class attributes (shared):")
print(f"  l1.gravity = {l1.gravity}  (same object? {l1.gravity is l2.gravity})")
print(f"  l2.gravity = {l2.gravity}")
print(f"  l3.gravity = {l3.gravity}")
print(f"  SoilLayer.gravity = {SoilLayer.gravity}")
print()

# Instance attributes are unique
print("Instance attributes (unique):")
print(f"  l1.depth = {l1.depth}")
print(f"  l2.depth = {l2.depth}")
print(f"  l3.depth = {l3.depth}")
print()

# The counter tracked all creations
print(f"Total layers created: {SoilLayer.layer_count}")
print()

# Effective unit weight (uses water_density class attribute)
wt = 2.5  # water table at 2.5m
print(f"Effective unit weights (water table at {wt}m):")
for layer in [l1, l2, l3]:
    gamma_eff = layer.effective_unit_weight(wt)
    above = "above WT" if layer.depth <= wt else "below WT"
    print(f"  {layer.soil_type:14s} depth={layer.depth}m ({above}): "
          f"\u03b3'={gamma_eff} kN/m\u00b3")
````

## Part 6: Building the Complete SoilLayer Class

Let's bring everything together into a fully functional `SoilLayer` class with all the methods we need for geotechnical analysis.

````python
# ─────────────────────────────────────────────────────
# COMPLETE SoilLayer CLASS — Full-featured version
# ─────────────────────────────────────────────────────

class SoilLayer:
    """
    Represents one soil layer in a borehole log.
    Stores properties and provides geotechnical analysis methods.
    """
    
    # Class attributes — physical constants
    gravity       = 9.81     # m/s²
    water_density = 9.81     # kN/m³
    atm_pressure  = 101.3    # kPa
    
    def __init__(self, soil_type, depth, N60, unit_weight):
        self.soil_type   = soil_type
        self.depth       = depth          # m below ground
        self.N60         = N60            # corrected SPT blow count
        self.unit_weight = unit_weight    # kN/m³
    
    # ---- Classification methods ----
    
    def classify_consistency(self):
        """SPT-based relative density classification (granular)."""
        if self.N60 < 4:   return "Very Loose"
        if self.N60 < 10:  return "Loose"
        if self.N60 < 30:  return "Medium Dense"
        if self.N60 < 50:  return "Dense"
        return "Very Dense"
    
    def is_cohesive(self):
        """Check if soil type is cohesive (clay/silt)."""
        cohesive_keywords = ["clay", "silt", "cohesive"]
        return any(kw in self.soil_type.lower() for kw in cohesive_keywords)
    
    # ---- Engineering property estimates ----
    
    def estimate_vs(self):
        """Shear wave velocity: Vs = 97 * N^0.314 (Imai & Tonouchi, 1982)."""
        if self.N60 <= 0:
            return 0.0
        return round(97 * (self.N60 ** 0.314), 1)
    
    def estimate_friction_angle(self):
        """Friction angle from SPT: phi = 27.1 + 0.3*N60 - 0.00054*N60^2 (Peck, 1974)."""
        if self.is_cohesive():
            return None  # not applicable for clay
        phi = 27.1 + 0.3 * self.N60 - 0.00054 * (self.N60 ** 2)
        return round(min(phi, 45.0), 1)  # cap at 45 degrees
    
    def overburden_stress(self):
        """Total vertical stress: sigma_v = gamma * depth."""
        return round(self.unit_weight * self.depth, 2)
    
    def effective_stress(self, water_table):
        """Effective vertical stress accounting for water table."""
        if self.depth <= water_table:
            return self.overburden_stress()
        u = self.water_density * (self.depth - water_table)  # pore pressure
        return round(self.overburden_stress() - u, 2)
    
    # ---- String representations ----
    
    def __str__(self):
        return (f"{self.soil_type} at {self.depth}m | "
                f"N60={self.N60}, \u03b3={self.unit_weight} kN/m\u00b3 | "
                f"{self.classify_consistency()}")
    
    def __repr__(self):
        return (f"SoilLayer('{self.soil_type}', {self.depth}, "
                f"{self.N60}, {self.unit_weight})")


# Demonstrate the complete class
layer = SoilLayer("Sand", 6.0, 22, 18.0)
wt = 3.0

print(f"Layer: {layer}")
print(f"  Shear wave velocity:  Vs = {layer.estimate_vs()} m/s")
print(f"  Friction angle:       \u03c6  = {layer.estimate_friction_angle()}\u00b0")
print(f"  Total stress:         \u03c3v = {layer.overburden_stress()} kPa")
print(f"  Effective stress:     \u03c3'v = {layer.effective_stress(wt)} kPa  (WT at {wt}m)")
print(f"  Cohesive soil?        {layer.is_cohesive()}")
````

## Part 7: Composition — Objects Containing Objects

The real power of OOP: a `BoreholeLog` can **contain** multiple `SoilLayer` objects.

```
BoreholeLog("BH-01")
  ├─ SoilLayer("Fill", 1.5, 5, 16.0)
  ├─ SoilLayer("Sand", 3.0, 12, 17.5)
  ├─ SoilLayer("Clay", 6.0, 8, 16.5)
  └─ SoilLayer("Gravel", 9.0, 35, 19.0)
```

This is called **composition** — building complex objects from simpler ones.

````python
# ─────────────────────────────────────────────────────
# BoreholeLog CLASS — Contains multiple SoilLayer objects
# ─────────────────────────────────────────────────────

class BoreholeLog:
    """
    Represents a complete borehole log containing multiple soil layers.
    Demonstrates composition: objects holding other objects.
    """
    
    def __init__(self, borehole_id, location="", water_table=None):
        self.borehole_id = borehole_id
        self.location    = location
        self.water_table = water_table    # m below ground
        self.layers      = []             # list of SoilLayer objects
    
    def add_layer(self, soil_type, depth, N60, unit_weight):
        """Create and add a SoilLayer to this borehole."""
        layer = SoilLayer(soil_type, depth, N60, unit_weight)
        self.layers.append(layer)
        return layer
    
    def total_depth(self):
        """Maximum depth of investigation."""
        if not self.layers:
            return 0.0
        return max(layer.depth for layer in self.layers)
    
    def find_weakest_layer(self):
        """Find the layer with the lowest N60 (weakest)."""
        if not self.layers:
            return None
        return min(self.layers, key=lambda l: l.N60)
    
    def find_strongest_layer(self):
        """Find the layer with the highest N60 (strongest)."""
        if not self.layers:
            return None
        return max(self.layers, key=lambda l: l.N60)
    
    def average_N60(self):
        """Average N60 across all layers."""
        if not self.layers:
            return 0.0
        return round(sum(l.N60 for l in self.layers) / len(self.layers), 1)
    
    def layers_below_water_table(self):
        """Return layers that are below the water table."""
        if self.water_table is None:
            return []
        return [l for l in self.layers if l.depth > self.water_table]
    
    def check_liquefaction_risk(self, earthquake_magnitude=6.5):
        """
        Simplified liquefaction screening:
        Risk exists if a sandy layer below WT has N60 < 15
        and depth < 20m (simplified criterion).
        """
        at_risk = []
        if self.water_table is None:
            return at_risk
        
        for layer in self.layers:
            is_sandy = not layer.is_cohesive()
            below_wt = layer.depth > self.water_table
            low_n    = layer.N60 < 15
            shallow  = layer.depth < 20.0
            
            if is_sandy and below_wt and low_n and shallow:
                at_risk.append(layer)
        
        return at_risk
    
    def generate_report(self):
        """Print a formatted borehole log report."""
        print(f"\n{'=' * 65}")
        print(f"  BOREHOLE LOG: {self.borehole_id}")
        if self.location:
            print(f"  Location: {self.location}")
        print(f"  Total depth: {self.total_depth()}m")
        wt_str = f"{self.water_table}m" if self.water_table else "Not encountered"
        print(f"  Water table: {wt_str}")
        print(f"{'=' * 65}")
        
        print(f"  {'Depth':>5} {'N60':>4} {'Soil':14s} {'\u03b3':>5} "
              f"{'Class':14s} {'Vs':>6} {'\u03c3v':>7}")
        print(f"  {'(m)':>5} {'':>4} {'':14s} {'kN/m\u00b3':>5} "
              f"{'':14s} {'(m/s)':>6} {'(kPa)':>7}")
        print("  " + "\u2500" * 60)
        
        for layer in self.layers:
            sigma = layer.effective_stress(self.water_table) if self.water_table else layer.overburden_stress()
            phi = layer.estimate_friction_angle()
            print(f"  {layer.depth:5.1f} {layer.N60:4d} {layer.soil_type:14s} "
                  f"{layer.unit_weight:5.1f} {layer.classify_consistency():14s} "
                  f"{layer.estimate_vs():6.1f} {sigma:7.1f}")
        
        print("  " + "\u2500" * 60)
        
        # Summary
        weakest = self.find_weakest_layer()
        print(f"\n  Summary:")
        print(f"    Average N60:     {self.average_N60()}")
        print(f"    Weakest layer:   {weakest.soil_type} at {weakest.depth}m (N60={weakest.N60})")
        print(f"    Layers below WT: {len(self.layers_below_water_table())}")
        
        # Liquefaction check
        at_risk = self.check_liquefaction_risk()
        if at_risk:
            print(f"\n  \u26a0 LIQUEFACTION RISK ({len(at_risk)} layer(s)):")
            for layer in at_risk:
                print(f"    - {layer.soil_type} at {layer.depth}m (N60={layer.N60})")
        else:
            print(f"\n  Liquefaction screening: No high-risk layers identified")
        
        print(f"{'=' * 65}")
    
    def __str__(self):
        return (f"BoreholeLog('{self.borehole_id}', "
                f"{len(self.layers)} layers, "
                f"depth={self.total_depth()}m)")
    
    def __repr__(self):
        return self.__str__()


print("BoreholeLog class defined. Ready for Part 8!")
````

## Part 8: Complete Application — Borehole Analysis

Let's build a real borehole log and run a full analysis using our two classes.

````python
# ═════════════════════════════════════════════════════
# STEP 1: Create the borehole and add layers
# ═════════════════════════════════════════════════════

bh = BoreholeLog("BH-01", location="Bridge Site, Station 5+200", water_table=2.5)

# Add layers from field data
bh.add_layer("Fill",         1.5,  5, 16.0)
bh.add_layer("Silty Sand",   3.0, 12, 17.5)
bh.add_layer("Sand",         6.0, 22, 18.0)
bh.add_layer("Soft Clay",    9.0,  8, 16.5)
bh.add_layer("Dense Sand",  12.0, 35, 19.0)
bh.add_layer("Gravel",      15.0, 55, 20.0)

print(f"Created: {bh}")
print(f"Number of layers: {len(bh.layers)}")
````

````python
# ═════════════════════════════════════════════════════
# STEP 2: Generate the full borehole report
# ═════════════════════════════════════════════════════

bh.generate_report()
````

````python
# ═════════════════════════════════════════════════════
# STEP 3: Query specific information from the borehole
# ═════════════════════════════════════════════════════

# Find weakest and strongest layers
weakest = bh.find_weakest_layer()
strongest = bh.find_strongest_layer()

print("Targeted queries on the borehole:")
print(f"  Weakest layer:   {weakest}")
print(f"  Strongest layer: {strongest}")
print(f"  Average N60:     {bh.average_N60()}")
print()

# Layers below water table
submerged = bh.layers_below_water_table()
print(f"Layers below water table ({bh.water_table}m):")
for layer in submerged:
    print(f"  - {layer}")
print()

# Liquefaction risk
at_risk = bh.check_liquefaction_risk()
print(f"Liquefaction risk layers:")
if at_risk:
    for layer in at_risk:
        print(f"  - {layer.soil_type} at {layer.depth}m (N60={layer.N60})")
else:
    print("  None identified")
````

````python
# ═════════════════════════════════════════════════════
# STEP 4: Compare multiple boreholes
# ═════════════════════════════════════════════════════

# Create a second borehole
bh2 = BoreholeLog("BH-02", location="Bridge Site, Station 5+220", water_table=3.0)
bh2.add_layer("Fill",        1.5,  4, 15.5)
bh2.add_layer("Loose Sand",  3.0,  7, 16.5)
bh2.add_layer("Silty Clay",  6.0, 10, 17.0)
bh2.add_layer("Sand",        9.0, 28, 18.5)
bh2.add_layer("Dense Sand", 12.0, 42, 19.5)

# Compare side by side
print("Borehole Comparison")
print("=" * 50)
for borehole in [bh, bh2]:
    weakest = borehole.find_weakest_layer()
    at_risk = borehole.check_liquefaction_risk()
    print(f"\n  {borehole.borehole_id} ({borehole.location})")
    print(f"    Layers:      {len(borehole.layers)}")
    print(f"    Total depth: {borehole.total_depth()}m")
    print(f"    Avg N60:     {borehole.average_N60()}")
    print(f"    Weakest:     {weakest.soil_type} at {weakest.depth}m (N60={weakest.N60})")
    print(f"    Liq. risk:   {len(at_risk)} layer(s)")
    if at_risk:
        for layer in at_risk:
            print(f"      - {layer.soil_type} at {layer.depth}m")
````

## Why OOP? — Compare With and Without

Let's see the same task done **without** OOP vs **with** OOP.

````python
# ─────────────────────────────────────────────────────
# WITHOUT OOP — parallel lists, error-prone
# ─────────────────────────────────────────────────────

print("WITHOUT OOP (parallel lists):")
print("-" * 40)

soils   = ["Fill",  "Sand",  "Clay"]
depths  = [1.5,     3.0,     6.0]
n_vals  = [5,       22,      8]
gammas  = [16.0,    18.0,    16.5]

# Find weakest — must manually track indices
min_n = min(n_vals)
min_i = n_vals.index(min_n)
print(f"  Weakest: {soils[min_i]} at {depths[min_i]}m (N60={n_vals[min_i]})")
print(f"  Problem: 4 separate lists must stay in sync!")
print()

# ─────────────────────────────────────────────────────
# WITH OOP — clean and intuitive
# ─────────────────────────────────────────────────────

print("WITH OOP (objects):")
print("-" * 40)

layers = [
    SoilLayer("Fill",  1.5,  5, 16.0),
    SoilLayer("Sand",  3.0, 22, 18.0),
    SoilLayer("Clay",  6.0,  8, 16.5),
]

# Find weakest — one readable line
weakest = min(layers, key=lambda l: l.N60)
print(f"  Weakest: {weakest}")
print(f"  Benefit: Each layer is ONE object. Data and behavior together.")
print()
print("OOP keeps related data together — harder to make mistakes.")
````

## Key Takeaways

1. **A class is a blueprint; an object is one thing built from it.** `SoilLayer` is the template; `SoilLayer("Sand", 6.0, 22, 18.0)` is one actual layer.

2. **`__init__` sets up each object with its data.** Every soil layer stores its own depth, N60, soil type, and unit weight — no parallel lists needed.

3. **Methods let objects DO things with their own data.** `layer.classify_consistency()` uses THAT layer's N60 — no need to pass it as an argument.

4. **`__str__` and `__repr__` make objects human-readable.** Without them, `print(layer)` shows a useless memory address.

5. **Class attributes are shared constants** (gravity, water density). Instance attributes are unique per object (depth, N60).

6. **Composition = objects inside objects.** A `BoreholeLog` holds multiple `SoilLayer` objects — this mirrors real engineering data naturally.

7. **OOP keeps data and behavior together,** making code easier to read, test, and extend compared to parallel lists and standalone functions.
