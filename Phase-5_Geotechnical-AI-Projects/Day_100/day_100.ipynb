## Part 5 — Three Highlight Code Snippets from the Journey

````python
# ─────────────────────────────────────────────────────────────────────────
# HIGHLIGHT 1: USCS Soil Classifier (Day 026)
# ─────────────────────────────────────────────────────────────────────────
# This was our first taste of ML: decision boundaries in geotechnical space.

def uscs_classifier(LL: float, PI: float, percent_fines: float) -> str:
    """USCS soil classification from Atterberg limits and fines content."""
    if percent_fines > 50:       # Fine-grained
        if LL >= 50:
            return 'CH' if PI > 0.73*(LL-20) else 'MH'
        else:
            if PI > 7 and PI > 0.73*(LL-20):
                return 'CL'
            elif PI < 4 or PI < 0.73*(LL-20):
                return 'ML'
            return 'CL-ML'
    else:                        # Coarse-grained (simplified)
        if percent_fines > 12:
            return 'SC' if PI > 7 else 'SM'
        return 'SW'  # further Cu/Cc check omitted for brevity

# Five representative soil types
samples = [
    (75, 42, 85, 'High-plasticity fat clay'),
    (38, 16, 78, 'Medium lean clay'),
    (28,  6, 72, 'Low-plasticity silt'),
    (22, 10, 30, 'Sandy clay'),
    (18,  4, 20, 'Silty sand'),
]
print('USCS Classification Demo (Day 026 highlight):')
print(f'{"Description":<35} {"LL":>4} {"PI":>4} {"Fines":>6} {"USCS":>6}')
print('-' * 60)
for LL, PI, pf, name in samples:
    symbol = uscs_classifier(LL, PI, pf)
    print(f'{name:<35} {LL:>4} {PI:>4} {pf:>5}%  {symbol:>6}')
````

````python
# ─────────────────────────────────────────────────────────────────────────
# HIGHLIGHT 2: Liquefaction Prediction Pipeline (Day 052)
# ─────────────────────────────────────────────────────────────────────────
# The most impactful model — safety-relevant, physically meaningful features.

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import GradientBoostingClassifier

def build_liquefaction_pipeline() -> Pipeline:
    """Gradient Boosting pipeline for liquefaction susceptibility.

    Input features: depth_m, N1_60, fines_pct, CSR_M, Vs30_mps
    Output: P(liquefaction) in [0, 1]
    """
    return Pipeline([
        ('scaler', StandardScaler()),
        ('clf',    GradientBoostingClassifier(
                       n_estimators=300,
                       max_depth=4,
                       learning_rate=0.05,
                       subsample=0.8,
                       random_state=42
        ))
    ])

print('Liquefaction pipeline (Day 052 highlight):')
pipe = build_liquefaction_pipeline()
print(pipe)
print()
print('Usage: pipe.fit(X_train, y_train)')
print('       proba = pipe.predict_proba(X_new)[:, 1]')
````

````python
# ─────────────────────────────────────────────────────────────────────────
# HIGHLIGHT 3: PINN Loss Function for 1D Consolidation (Day 087)
# ─────────────────────────────────────────────────────────────────────────
# Our most advanced work — embedding Terzaghi's PDE into the training loss.

PINN_CODE = '''
# Physics-Informed Neural Network — 1D Terzaghi Consolidation (Day 087)
# Equation:  du/dt = Cv * d²u/dz²
# u = excess pore water pressure [kPa]
# z = depth [m],  t = time [s],  Cv = coefficient of consolidation [m²/s]

import torch
import torch.nn as nn

class ConsolidationPINN(nn.Module):
    def __init__(self, Cv: float = 1e-7):
        super().__init__()
        self.Cv = Cv
        self.net = nn.Sequential(
            nn.Linear(2, 64), nn.Tanh(),  # input: [z, t]
            nn.Linear(64, 64), nn.Tanh(),
            nn.Linear(64, 64), nn.Tanh(),
            nn.Linear(64, 1)              # output: u(z, t)
        )

    def forward(self, z, t):
        return self.net(torch.cat([z, t], dim=1))

    def pde_residual(self, z, t):
        """Residual of du/dt - Cv * d²u/dz² = 0."""
        z.requires_grad_(True)
        t.requires_grad_(True)
        u = self.forward(z, t)

        # First derivatives
        u_t  = torch.autograd.grad(u.sum(), t, create_graph=True)[0]
        u_z  = torch.autograd.grad(u.sum(), z, create_graph=True)[0]

        # Second derivative
        u_zz = torch.autograd.grad(u_z.sum(), z, create_graph=True)[0]

        return u_t - self.Cv * u_zz   # should be zero everywhere


def total_pinn_loss(model, z_data, t_data, u_data,
                    z_pde, t_pde, z_bc, t_bc):
    """Composite PINN loss: data + PDE residual + boundary conditions."""
    # Data loss (measured pore pressures)
    u_pred   = model(z_data, t_data)
    loss_data = nn.MSELoss()(u_pred, u_data)

    # PDE residual loss (physics)
    residual  = model.pde_residual(z_pde, t_pde)
    loss_pde  = (residual ** 2).mean()

    # Boundary condition: u(z=H, t) = 0  (free-draining top)
    u_bc      = model(z_bc, t_bc)
    loss_bc   = nn.MSELoss()(u_bc, torch.zeros_like(u_bc))

    return loss_data + 0.1 * loss_pde + 0.1 * loss_bc
'''

print('PINN Loss Function for 1D Consolidation (Day 087 highlight):')
print('=' * 65)
print(PINN_CODE)
````
