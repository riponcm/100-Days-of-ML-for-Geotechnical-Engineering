## Our Geotechnical Problem

We have SPT (Standard Penetration Test) data from a site investigation. The data lives in lists and Pandas DataFrames, but numbers alone do not tell the story. We need **plots**:

- A **depth profile** of N-values — the classic borehole log that every geotechnical engineer reads daily
- A **scatter plot** correlating N-values to shear wave velocity (Vs)
- A **bar chart** showing how many samples of each soil type were found
- A **histogram** revealing the statistical distribution of blow counts

Matplotlib is Python’s foundational plotting library. Every ML visualization library (Seaborn, Plotly, even Pandas `.plot()`) is built on top of or inspired by Matplotlib.

## Part 2: Line Plot — SPT N-value Depth Profile

The **depth profile** is the most common plot in geotechnical engineering. Convention:
- **x-axis**: N-value (blow count)
- **y-axis**: Depth (increasing downward — inverted)

Key functions:
| Function | Purpose |
|----------|--------|
| `plt.figure(figsize=(w, h))` | Set figure size in inches |
| `plt.plot(x, y)` | Draw a line connecting points |
| `plt.xlabel()` / `plt.ylabel()` | Label the axes |
| `plt.title()` | Add a title |
| `plt.gca().invert_yaxis()` | Flip y-axis (depth downward) |
| `plt.grid(True)` | Add gridlines |

````python
# ─────────────────────────────────────────────────────
# LINE PLOT — Professional SPT depth profile
# ─────────────────────────────────────────────────────

depth_m = [1.5, 3.0, 4.5, 6.0, 7.5, 9.0, 10.5, 12.0]
n_value = [5, 8, 12, 18, 28, 35, 50, 60]

# Create figure with explicit size
plt.figure(figsize=(6, 8))

# Plot N-value (x) vs depth (y) with markers
plt.plot(n_value, depth_m, marker='o', color='steelblue',
         linewidth=1.5, markersize=6, label='BH-01')

# Labels and title
plt.xlabel('SPT N-value (blows/300mm)', fontsize=12)
plt.ylabel('Depth (m)', fontsize=12)
plt.title('SPT N-value Profile \u2014 BH-01', fontsize=14, fontweight='bold')

# Invert y-axis so depth increases downward
plt.gca().invert_yaxis()

# Grid and legend
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='lower right', fontsize=10)

# Tight layout prevents label clipping
plt.tight_layout()
plt.show()

print("Now depth increases downward \u2014 standard geotechnical convention!")
````

````python
# ─────────────────────────────────────────────────────
# MULTI-BOREHOLE LINE PLOT — Overlay profiles
# ─────────────────────────────────────────────────────

# Three boreholes from same site
depth_m  = [1.5, 3.0, 4.5, 6.0, 7.5, 9.0, 10.5, 12.0]
n_bh01   = [5, 8, 12, 18, 28, 35, 50, 60]
n_bh02   = [3, 6, 10, 15, 22, 30, 45, 55]
n_bh03   = [7, 12, 16, 22, 32, 40, 52, 65]

plt.figure(figsize=(7, 8))

# Plot each borehole with different style
plt.plot(n_bh01, depth_m, 'o-',  color='steelblue',   linewidth=1.5, label='BH-01')
plt.plot(n_bh02, depth_m, 's--', color='darkorange',  linewidth=1.5, label='BH-02')
plt.plot(n_bh03, depth_m, '^:',  color='forestgreen', linewidth=1.5, label='BH-03')

# Geotechnical convention: depth downward
plt.gca().invert_yaxis()

# Labels
plt.xlabel('SPT N-value (blows/300mm)', fontsize=12)
plt.ylabel('Depth (m)', fontsize=12)
plt.title('SPT Profiles \u2014 Site Comparison', fontsize=14, fontweight='bold')

# Reference line at N=30 (dense boundary)
plt.axvline(x=30, color='red', linestyle=':', alpha=0.5, label='Dense boundary (N=30)')

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower right', fontsize=10)
plt.tight_layout()
plt.show()

print("Three boreholes overlaid \u2014 BH-03 is the stiffest profile.")
print("Format string 'o-'  = circle marker + solid line")
print("Format string 's--' = square marker + dashed line")
print("Format string '^:'  = triangle marker + dotted line")
````

## Part 3: Scatter Plot — N-value vs Shear Wave Velocity

Scatter plots show the **relationship between two variables** without connecting lines. In geotechnical engineering, correlations between field tests (SPT N) and lab/geophysical measurements (Vs) are critical.

The empirical relationship: **Vs = 97 × N^0.314** (Imai & Tonouchi, 1982)

| `plt.scatter()` parameter | Purpose |
|--------------------------|--------|
| `c=` | Color of points (or array for colormap) |
| `s=` | Size of points |
| `alpha=` | Transparency (0=invisible, 1=solid) |
| `cmap=` | Colormap when `c=` is numeric array |

````python
# ─────────────────────────────────────────────────────
# SCATTER PLOT — N-value vs Shear Wave Velocity (Vs)
# ─────────────────────────────────────────────────────

# Simulated field data: N-values and measured Vs
np.random.seed(42)
n_field = np.array([3, 5, 7, 8, 10, 12, 15, 18, 20, 22,
                    25, 28, 30, 33, 35, 38, 40, 45, 50, 55, 60])

# Vs from empirical relation + noise (real data is never perfect)
vs_measured = 97 * (n_field ** 0.314) + np.random.normal(0, 15, len(n_field))

# Empirical curve for reference
n_curve = np.linspace(1, 65, 100)
vs_curve = 97 * (n_curve ** 0.314)

plt.figure(figsize=(8, 6))

# Scatter: measured data
plt.scatter(n_field, vs_measured, c='steelblue', s=60, alpha=0.8,
            edgecolors='navy', label='Field measurements')

# Line: empirical correlation
plt.plot(n_curve, vs_curve, 'r--', linewidth=1.5,
         label='Vs = 97 N$^{0.314}$ (Imai & Tonouchi)')

plt.xlabel('SPT N-value', fontsize=12)
plt.ylabel('Shear Wave Velocity, Vs (m/s)', fontsize=12)
plt.title('N-value vs Shear Wave Velocity Correlation', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Scatter shows individual measurements; the line is the empirical fit.")
print("Real data always has scatter around the theoretical curve.")
````

````python
# ─────────────────────────────────────────────────────
# SCATTER WITH COLORMAP — Color points by depth
# ─────────────────────────────────────────────────────

# Each point comes from a specific depth
depth_values = np.array([1.5, 2.0, 3.0, 3.5, 4.0, 4.5, 5.0, 6.0, 6.5, 7.0,
                         7.5, 8.0, 9.0, 9.5, 10.0, 10.5, 11.0, 12.0, 13.0, 14.0, 15.0])

plt.figure(figsize=(8, 6))

# Color by depth using a colormap
scatter = plt.scatter(n_field, vs_measured, c=depth_values, s=80,
                      cmap='viridis_r', alpha=0.9, edgecolors='black', linewidths=0.5)

# Add colorbar to show depth scale
cbar = plt.colorbar(scatter)
cbar.set_label('Depth (m)', fontsize=11)

plt.xlabel('SPT N-value', fontsize=12)
plt.ylabel('Shear Wave Velocity, Vs (m/s)', fontsize=12)
plt.title('N vs Vs \u2014 Colored by Depth', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Darker points = shallower; lighter points = deeper.")
print("The colorbar adds a third dimension to the scatter plot.")
````

## Part 4: Bar Chart — Soil Type Distribution

Bar charts show **categorical data** — perfect for counting how many samples of each soil type appear in a site investigation.

| `plt.bar()` parameter | Purpose |
|----------------------|--------|
| `x` | Category positions or labels |
| `height` | Bar heights (counts) |
| `color=` | Bar colors |
| `edgecolor=` | Bar outline color |
| `width=` | Bar width (default 0.8) |

````python
# ─────────────────────────────────────────────────────
# BAR CHART — Soil type distribution
# ─────────────────────────────────────────────────────

# Soil types found across all boreholes
soil_types   = ['Sand', 'Silty Sand', 'Clay', 'Gravel', 'Fill', 'Silt', 'Bedrock']
sample_count = [12, 8, 6, 5, 3, 3, 2]
soil_colors  = ['#F4D03F', '#D4AC0D', '#A0522D', '#808080', '#8B4513', '#BDB76B', '#696969']

plt.figure(figsize=(9, 5))

bars = plt.bar(soil_types, sample_count, color=soil_colors,
               edgecolor='black', linewidth=0.8)

# Add count labels on top of each bar
for bar, count in zip(bars, sample_count):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.2,
             str(count), ha='center', va='bottom', fontweight='bold', fontsize=11)

plt.xlabel('Soil Type', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)
plt.title('Soil Type Distribution \u2014 Site Investigation', fontsize=14, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print(f"Total samples: {sum(sample_count)}")
print(f"Most common: {soil_types[0]} ({sample_count[0]} samples)")
````

````python
# ─────────────────────────────────────────────────────
# HORIZONTAL BAR CHART — Average N-value by soil type
# ─────────────────────────────────────────────────────

soil_types_h = ['Fill', 'Silt', 'Clay', 'Silty Sand', 'Sand', 'Gravel', 'Bedrock']
avg_n_value  = [4, 7, 9, 14, 25, 38, 58]

plt.figure(figsize=(8, 5))

# barh = horizontal bars
plt.barh(soil_types_h, avg_n_value, color='steelblue',
         edgecolor='navy', height=0.6)

# Vertical reference lines for density classes
plt.axvline(x=10, color='orange', linestyle=':', alpha=0.7, label='Loose/Medium (N=10)')
plt.axvline(x=30, color='red', linestyle=':', alpha=0.7, label='Medium/Dense (N=30)')

plt.xlabel('Average SPT N-value', fontsize=12)
plt.ylabel('Soil Type', fontsize=12)
plt.title('Average N-value by Soil Type', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=10)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Horizontal bars are better when category labels are long.")
````

## Part 5: Histogram — N-value Frequency Distribution

A histogram shows **how frequently values fall in each range** (bin). Unlike bar charts (categorical), histograms work with **continuous numeric data**.

In geotechnical engineering, we use histograms to:
- Understand the distribution of blow counts across a site
- Check if data is skewed (common for N-values: many low, few very high)
- Identify outliers or unusual clusters

| `plt.hist()` parameter | Purpose |
|-----------------------|--------|
| `bins=` | Number of bins or bin edges |
| `edgecolor=` | Bin outline color |
| `alpha=` | Transparency for overlapping histograms |
| `density=True` | Normalize to show probability instead of count |

````python
# ─────────────────────────────────────────────────────
# HISTOGRAM — N-value frequency distribution
# ─────────────────────────────────────────────────────

# All N-values from 3 boreholes combined
n_all_boreholes = np.concatenate([
    np.array([5, 8, 12, 18, 28, 35, 50, 60]),    # BH-01
    np.array([3, 6, 10, 15, 22, 30, 45, 55]),    # BH-02
    np.array([7, 12, 16, 22, 32, 40, 52, 65]),   # BH-03
])

plt.figure(figsize=(8, 5))

# Histogram with custom bins
counts, bins, patches = plt.hist(n_all_boreholes, bins=10,
                                  color='steelblue', edgecolor='black',
                                  alpha=0.8)

plt.xlabel('SPT N-value', fontsize=12)
plt.ylabel('Frequency (count)', fontsize=12)
plt.title('N-value Distribution \u2014 All Boreholes', fontsize=14, fontweight='bold')

# Add mean and median lines
mean_n = np.mean(n_all_boreholes)
median_n = np.median(n_all_boreholes)
plt.axvline(mean_n, color='red', linestyle='-', linewidth=2, label=f'Mean = {mean_n:.1f}')
plt.axvline(median_n, color='orange', linestyle='--', linewidth=2, label=f'Median = {median_n:.1f}')

plt.legend(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print(f"N-values range: {n_all_boreholes.min()} to {n_all_boreholes.max()}")
print(f"Mean: {mean_n:.1f}, Median: {median_n:.1f}")
print("Right-skewed: many low N-values (shallow), few very high (deep rock).")
````

````python
# ─────────────────────────────────────────────────────
# OVERLAPPING HISTOGRAMS — Compare borehole distributions
# ─────────────────────────────────────────────────────

n_bh01 = [5, 8, 12, 18, 28, 35, 50, 60]
n_bh02 = [3, 6, 10, 15, 22, 30, 45, 55]
n_bh03 = [7, 12, 16, 22, 32, 40, 52, 65]

common_bins = np.arange(0, 70, 10)  # 0, 10, 20, ..., 60

plt.figure(figsize=(8, 5))

# Overlapping histograms with transparency
plt.hist(n_bh01, bins=common_bins, alpha=0.5, color='steelblue',
         edgecolor='black', label='BH-01')
plt.hist(n_bh02, bins=common_bins, alpha=0.5, color='darkorange',
         edgecolor='black', label='BH-02')
plt.hist(n_bh03, bins=common_bins, alpha=0.5, color='forestgreen',
         edgecolor='black', label='BH-03')

plt.xlabel('SPT N-value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('N-value Distribution by Borehole', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Overlapping histograms use alpha (transparency) to compare distributions.")
print("All three boreholes have similar spread patterns.")
````

## Part 6: Figure Customization Basics

Before we move to the complete application, let’s master the essential customization tools:

| Customization | Code |
|--------------|------|
| Figure size | `plt.figure(figsize=(width, height))` |
| Font sizes | `fontsize=12` in labels/titles |
| Colors | Named colors, hex codes, or RGB tuples |
| Line styles | `'-'` solid, `'--'` dashed, `':'` dotted, `'-.'` dash-dot |
| Markers | `'o'` circle, `'s'` square, `'^'` triangle, `'D'` diamond |
| Grid | `plt.grid(True, linestyle='--', alpha=0.5)` |
| Axis limits | `plt.xlim(0, 70)`, `plt.ylim(0, 15)` |

````python
# ─────────────────────────────────────────────────────
# CUSTOMIZATION DEMO — Colors, markers, line styles
# ─────────────────────────────────────────────────────

depth_m = np.array([1.5, 3.0, 4.5, 6.0, 7.5, 9.0, 10.5, 12.0])
n_value = np.array([5, 8, 12, 18, 28, 35, 50, 60])

# Compute derived values
vs_imai = 97 * (n_value ** 0.314)               # Imai & Tonouchi
vs_ohta = 85.35 * (n_value ** 0.348)             # Ohta & Goto

plt.figure(figsize=(8, 6))

# Different line styles, markers, and colors
plt.plot(n_value, vs_imai, 'o-', color='#2196F3', linewidth=2,
         markersize=8, label='Imai & Tonouchi (1982)')
plt.plot(n_value, vs_ohta, 's--', color='#FF5722', linewidth=2,
         markersize=8, label='Ohta & Goto (1978)')

# Axis limits and ticks
plt.xlim(0, 65)
plt.ylim(100, 400)

# Labels with units
plt.xlabel('SPT N-value (blows/300mm)', fontsize=12)
plt.ylabel('Shear Wave Velocity, Vs (m/s)', fontsize=12)
plt.title('Vs Empirical Correlations Comparison', fontsize=14, fontweight='bold')

plt.legend(fontsize=11, loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Named colors: 'steelblue', 'red', 'forestgreen'")
print("Hex colors:   '#2196F3', '#FF5722'")
print("Markers:      'o' circle, 's' square, '^' triangle, 'D' diamond")
print("Lines:        '-' solid, '--' dashed, ':' dotted, '-.' dash-dot")
````

## Part 7: Complete Application — Professional SPT Borehole Log

Now we combine everything to create a **professional SPT borehole log plot** — the kind of plot that appears in every geotechnical investigation report:

- Depth on y-axis (inverted, increasing downward)
- N-value on x-axis with marker at each test depth
- Water table indicator
- Density classification zones
- Soil type annotations
- Professional formatting (grid, labels, title block)

````python
# ═════════════════════════════════════════════════
# COMPLETE: PROFESSIONAL SPT BOREHOLE LOG PLOT
# ═════════════════════════════════════════════════

# Borehole data
bh_id = 'BH-01'
water_table_m = 3.2

depth_m    = [1.5, 3.0, 4.5, 6.0, 7.5, 9.0, 10.5, 12.0, 13.5, 15.0]
n_value    = [5, 8, 12, 18, 28, 35, 42, 50, 55, 60]
soil_type  = ['Fill', 'Silty Sand', 'Sand', 'Clayey Sand', 'Sand',
              'Gravel', 'Gravel', 'Weathered Rock', 'Weathered Rock', 'Bedrock']

# Create figure
fig, ax = plt.subplots(figsize=(8, 10))

# ---- N-value profile (main line) ----
ax.plot(n_value, depth_m, 'o-', color='steelblue', linewidth=2,
        markersize=7, markerfacecolor='steelblue', markeredgecolor='navy',
        zorder=5, label='SPT N-value')

# ---- Density classification bands ----
ax.axvspan(0, 4, alpha=0.08, color='red', label='Very Loose (N<4)')
ax.axvspan(4, 10, alpha=0.08, color='orange', label='Loose (4-10)')
ax.axvspan(10, 30, alpha=0.08, color='yellow', label='Medium Dense (10-30)')
ax.axvspan(30, 50, alpha=0.08, color='lightgreen', label='Dense (30-50)')
ax.axvspan(50, 70, alpha=0.08, color='green', label='Very Dense (50+)')

# ---- Water table indicator ----
ax.axhline(y=water_table_m, color='deepskyblue', linestyle='-.',
           linewidth=2, alpha=0.8, label=f'Water Table ({water_table_m}m)')

# ---- Soil type annotations ----
for d, n, soil in zip(depth_m, n_value, soil_type):
    ax.annotate(f'  {soil}', xy=(n, d), fontsize=8,
                va='center', ha='left', color='dimgray')

# ---- Axis formatting ----
ax.invert_yaxis()                          # Depth increases downward
ax.set_xlim(0, 70)
ax.set_ylim(16, 0)                         # 0 at top, max depth at bottom
ax.set_xlabel('SPT N-value (blows/300mm)', fontsize=13, fontweight='bold')
ax.set_ylabel('Depth (m)', fontsize=13, fontweight='bold')
ax.set_title(f'Standard Penetration Test Log \u2014 {bh_id}',
             fontsize=15, fontweight='bold', pad=15)

# ---- Grid and legend ----
ax.grid(True, linestyle='--', alpha=0.4)
ax.legend(loc='lower right', fontsize=8, framealpha=0.9)

plt.tight_layout()
plt.show()

print(f"Borehole: {bh_id}")
print(f"Depth range: {min(depth_m)}\u2013{max(depth_m)} m")
print(f"N-value range: {min(n_value)}\u2013{max(n_value)}")
print(f"Water table: {water_table_m} m")
print("\nThis is the standard format used in geotechnical investigation reports.")
````

## Key Takeaways

1. **`plt.plot()`** creates line plots — perfect for SPT depth profiles where we connect N-values at each test depth.

2. **`plt.scatter()`** shows individual data points — ideal for correlation plots like N vs Vs where each point is an independent measurement.

3. **`plt.bar()`** displays categorical comparisons — soil type counts, average properties by classification.

4. **`plt.hist()`** reveals frequency distributions — understanding how N-values are spread across a site.

5. **`plt.gca().invert_yaxis()`** is essential for geotechnical plots — depth must always increase downward.
